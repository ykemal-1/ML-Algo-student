## KNN
Implementeer KNN in Python, gebruik de code snippet hieronder als startpunt.

```python
import numpy as np

class KNNClassifier:
    def __init__(self, k=3):
        self.k = k # ZELF: aantal dichtstbijzijnde huizen dat meetelt
        self.X_train = None
        self.y_train = None

    def fit(self, X, y):
        #todo

         # ZELF: bewaar de bekende huizen (X) en hun prijzen (y)
        self.X_train = X
        self.y_train = y

    def euclidean_distance(self, x1, x2): # deze Euclidische afstand krijg je cadeau

        # ZELF: één getal dat aangeeft hoe ver twee huizen uit elkaar liggen (Het is gwn een formule)
        return np.sqrt(np.sum((x1 - x2)**2))

    def predict(self, X):
        # ZELF: _predict wordt één keer uitgevoerd per nieuw huis
        predictions = [self._predict(x) for x in X]
        return np.array(predictions)

    def _predict(self, x):
        #todo
        #todo: 1 item voorspellen
        # bereken voor X de afstand met alle items van X_train
        # sorteren volgens van klein naar groot volgens die afstand
        # de eerste k labels (y data): neem daarvan het gemiddelde
        # gebruik np.argsort, np.mean

        # ZELF: x is één nieuw huis; bereken de afstand tot elk bekend huis
        distances = [
            self.euclidean_distance(x, house)
              for house in self.X_train
              ]

        # ZELF: argsort geeft de posities van dichtbij naar veraf;
        # neem daarvan alleen de eerste k

        nearest_indices = np.argsort(distances)[:self.k]

        # ZELF: gebruik die posities om de bijbehorende prijzen te vinden
        nearest_prices = self.y_train[nearest_indices]

        # ZELF: de gemiddelde prijs van de buren is onze voorspelling
        return np.mean(nearest_prices)

        

# Sample dataset

# ZELF: elke rij is een bekend huis; y_train bevat de prijs van diezelfde rij
X_train = np.array([
    [0, 0, 1000, 2, 1],
    [1, 1, 1500, 3, 2],
    [2, 2, 1200, 2, 1],
    [3, 3, 1800, 4, 2],
    [4, 4, 2000, 3, 2]
])

y_train = np.array([200000, 250000, 220000, 280000, 300000])


# ZELF: voor deze twee nieuwe huizen kennen we de prijs nog niet
X_test = np.array([
    [5, 5, 1600, 3, 1],
    [2, 3, 1300, 2, 1]
])

knn = KNNClassifier(k=3)

# ZELF: [:, 1:] betekent alle rijen, maar zonder de eerste kolom
knn.fit(X_train[:, 1:], y_train)  # Ignoring the x-y coordinates for simplicity
predictions = knn.predict(X_test[:, 1:])
print("Predicted house prices:", predictions)
```

### Iris Flowers dataset

- Haal de dataset van de iris flowers binnen. Ze zit ingebakken in sci-kit-learn
- Gebruik de sci-kit learn library om een KNN classifier te trainen.
- Vergeet niet je algoritme te scoren

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Dataset inladen
iris = load_iris()
X = iris.data  # Features
y = iris.target  # de labels (bloemsoorten)
#print(iris.DESCR)
print(iris.keys())
iris.target_names
```

```python
# zoek op hoe je data kan train-test splitten
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=20, 
    random_state=42)
```

```python
# Gebruik een standardscaler op de X_train, X-test data

# Maak een StandardScaler aan
scaler = StandardScaler()

# Bereken gemiddelde en spreiding op X_train en schaal X_train
X_train_scaled = scaler.fit_transform(X_train)

# Schaal X_test met DEZELFDE instellingen
X_test_scaled = scaler.transform(X_test)
```

```python
# Gebruik een KNeighborsclassifier van de sklearn library. Fit deze. Maak voorspellingen.

# Maak een KNN-classifier die naar de 3 dichtstbijzijnde bloemen kijkt
knn = KNeighborsClassifier(n_neighbors=3)

# Leer met de trainingsmetingen (X) en de juiste bloemsoorten (y)
knn.fit(X_train_scaled, y_train)

# Voorspel de bloemsoorten van de testbloemen
y_pred = knn.predict(X_test_scaled)

print(y_pred)
```

```python
# Bereken de accuracy.

accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

```python
# maak een classificatie report

report = classification_report(
    y_test,
    y_pred,
    target_names=iris.target_names
)

print(report)

print(len(X_test), len(y_test), len(y_pred))
```

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd
data = pd.read_csv('./data/Mall_Customers.csv')
```

```python
data.head()
X_numerical = data[['Age', 'Annual Income (k$)']]
X_catgeorical = data[['Gender']]
y = data[['Spending Score (1-100)']]
```

Gebruik one-hot encoding om de categorische variabelen om te zetten naar categorische variabelen.

```python
#todo

# Maak de encoder. sparse_output=False geeft een gewone array terug.
encoder = OneHotEncoder(sparse_output=False)

# Zet Gender om naar twee kolommen: Gender_Female en Gender_Male
gender_encoded = encoder.fit_transform(X_catgeorical)

# Maak er een DataFrame van met duidelijke kolomnamen
gender_df = pd.DataFrame(
    gender_encoded,
    columns=encoder.get_feature_names_out(['Gender']),
    index=data.index
)

# Voeg de nieuwe Gender-kolommen bij de numerieke kolommen
X = pd.concat([X_numerical, gender_df], axis=1)

print(X.head())
```

```python
# terug samenvoegen met numerieke variabelen

# Voeg de numerieke kolommen en de nieuwe Gender-kolommen naast elkaar
X = pd.concat([X_numerical, gender_df], axis=1)

print(X.head())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error, r2_score

# Split the data into training and testing sets

# ZELF: Verdeel de klanten: 80% om te trainen, 20% om te testen
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

# ZELF: schaal X, omdat KNN afstanden tussen klanten berekent
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Maak KNN regressor

# ZELF: voorspel de score op basis van de 5 dichtstbijzijnde klanten
knn = KNeighborsRegressor(n_neighbors=5)



# Train model

# ZELF: Train het model met klanten en hun bekende Spending Scores
knn.fit(X_train_scaled, y_train)

# Voorspelling

# ZELF: voorspel de Spending Scores van de testklanten
y_pred = knn.predict(X_test_scaled)

# Evaluatie met mean squared error en r2 score

# ZELF: vergelijk de voorspelde scores met de echte scores
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("MSE:", mse)
print("R²:", r2)
```
