# Oefening 1 — SVM op Breast Cancer

- Splits (test_size=0.25, random_state=10), normaliseer met StandardScaler, train SVC(random_state=10).

```python
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

X, y = load_breast_cancer(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=10)
scaler = StandardScaler().fit(X_train)
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
clf = SVC(random_state=10).fit(X_train_scaled, y_train)
y_pred_breast_cancer = clf.predict(X_test_scaled)
accuracy_breast_cancer = accuracy_score(y_test, y_pred_breast_cancer)
print(accuracy_breast_cancer)
```

**antwoord_scaling**: B — afstand/inwendig-productgebaseerd; schaalverschillen domineren anders de marge.

```python
antwoord_scaling = "B"
```

# Oefening 2 — SVM op Digits (pipeline)

- Pipeline met StandardScaler + SVC(random_state=10) op de digits dataset.

```python
import numpy as np
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.pipeline import make_pipeline
from sklearn.metrics import accuracy_score

digits = datasets.load_digits()
X, y = digits.data, digits.target
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=10)
pipe = make_pipeline(StandardScaler(), SVC(random_state=10)).fit(X_train, y_train)
y_pred_digits = pipe.predict(X_test)
accuracy_digits = accuracy_score(y_test, y_pred_digits)
print(accuracy_digits)
```

**antwoord_kernel**: C — rbf presteert doorgaans het best voor digits.

```python
antwoord_kernel = "A"
```

# Oefening 3 — SVM op Wine (meerklass)

- Pipeline met StandardScaler + SVC(random_state=10) op de wine dataset.

```python
import numpy as np
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.pipeline import make_pipeline
from sklearn.metrics import accuracy_score

X, y = load_wine(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=10)
pipe = make_pipeline(StandardScaler(), SVC(random_state=10)).fit(X_train, y_train)
accuracy_wine = accuracy_score(y_test, pipe.predict(X_test))
print(accuracy_wine)
```

# Oefening 4 — SVR (regressie)

- Pipeline met StandardScaler + SVR(kernel='poly', degree=2, C=0.01, epsilon=0.1).

```python
import numpy as np
from sklearn.svm import SVR
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error

np.random.seed(42)
X = 2 * np.random.rand(50, 1) - 1
y = 0.2 + 0.1 * X[:, 0] + 0.5 * X[:, 0] ** 2 + np.random.randn(50) / 10

model = make_pipeline(
    StandardScaler(),
    SVR(kernel="poly", degree=2, C=0.01, epsilon=0.1)).fit(X, y)
y_pred_svr = model.predict(X)
mse_svr = mean_squared_error(y, y_pred_svr)
print(mse_svr)
```
